## **Introduction to Algorithms**

### **What is an Algorithm?**

An algorithm is a finite sequence of well-defined instructions (Unambiguous) that can be mechanically executed (repetition -> automation) to solve a computanional problem or perform a task.  
> Un algoritmo es una secuencia finita de instrucciones bien definidas (sin ambigüedades) que pueden ejecutarse mecánicamente (repetición → automatización) para resolver un problema computacional o realizar una tarea

**Key Components**
- **Input:** Zero or more quantities supplied externally
- **Output:** One or more quantities produced
- **Instructions:** Precise and unambiguous steps
- **Execution:** Can be performed mechanically

**Remember:** An algorithm is a method for solving problems, not the implementation itself! (Algorithm ≠ Code)  
> Recuerda: Un algoritmo es un método para resolver problemas, ¡no la implementación en sí! (Algoritmo ≠ Código)

### **Algorithm Characteristics**

**Esential Properties**
- **Finiteness:** Must terminate after finite number of steps
- **Definiteness:** Each step must be precisely defined
- **Input:** Zero or more input values
- **Output:** One or more output values
- **Effectiveness:** Operations must be basic and feasible

    **_Example_**
    1. Boil water (Definite Action)
    2. Add coffee grounds (Precise Amount)
    3. Wait 4 minutes (Finite Time)
    4. Pour into cup (Effective Operation)

### **Algorithm vs. Program vs. Process**

**Algorithm**
- Abstract Concept
- Language Independent
- Mathematical Description
- Design Phase

**Program**
- Concrete Implementation
- Language Specific
- Code Representation
- Implementation Phase

**Process**
- Runtime Execution
- System Dependent
- Active Computation
- Execution Phase

    **_Example_**

    Finding the maximum number:
    - **Algorithm:** ”Compare each element with current maximum”
    - **Program:** Python code implementing the comparison
    - **Process:** Program running in memory, using CPU cycles

### **Types of Algorithmic Thinking**

- **Sequential Thinking:** Step-by-step execution
- **Conditional Thinking:** Decision-based branching
- **Iterative Thinking:** Repetitive operations
- **Recursive Thinking:** Self-referential solutions
- **Parallel Thinking:** Concurrent execution

    **_Example (Daily Life)_**
    - Getting dressed (Sequential)
    - Choosing route to university (Conditional)
    - Studying until understanding (Iterative)
    - Recalling a recipe (Recursive)
    - Cooking multiple dishes (Parallel)

### **Study Case: Euclidean GCD Algorithm**

**Greatest Common Divisor Problem**
- **Input:** Two positive integers a and b
- **Output:** The largest integer that divides both a and b
> - **Entrada:** Dos números enteros positivos a y b
> - **Salida:** El mayor número entero que divide tanto a a como a b

**Algorithm 2: Euclidean GCD Algorithm**  
- **Input:** Integers a, b where a ≥ b > 0

**while** b ̸= 0 **do**  
r ← a mod b  
a ← b  
b ← r  
**end while**  
**return** a  

```python
def euclidean_gcd(a: int, b: int) -> int:
    """
    This function implements the euclidean algorithms
    to get the greater common divisor of two numbers.

    Args:
        a(int): First number in the algorithm. It is expected to be the greater one.
        b(int): Second number in the algorithm.

    Returns:
        An integer with the greater common divisor of the two numbers.
    """
    if b > a:
        a, b = b, a

    while b != 0:
        r = a % b
        a = b
        b = r
    
    return a

print(euclidean_gcd(12,15))
```

```python
from typing import List
def bubble_sort(a: List) -> List:
    n = len(a)
    for i in range (n - 1):
        for j in range (n - 1 - i):
            if a[j] > a[j + 1]:
                swap = a[j]
                a[j] = a[j + 1]
                a[j + 1] = swap
    return a

from random import randint
a = [randint(1, 100000) for _ in range(10000)]
print(a)
a = bubble_sort(a)
print(a)
```

```python
def selection_sort(a: List) -> List:
    n = len(a)
    for i in range (n - 1):
        min_index = i
        for j in range (i + 1, n):
            if a[j] < a [min_index]:
                min_index = j

        swap = a[i]
        a[i] = a[min_index]
        a[min_index] = swap
    return a

from random import randint
a = [randint(1, 100000) for _ in range(10000)]
print(a)
a = selection_sort(a)
print(a)
```
